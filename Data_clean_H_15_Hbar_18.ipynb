```python
import pandas as pd
import glob
```

```python
all_loops = []
for fn in glob.glob("plugin_*_subsetsize_210_*.csv"):
    print(fn)
    df = pd.read_csv(fn)
    # drop all the padded rows:
    df = df[df["v_pi_fqe"].notna()].copy()

    # split your CI columns into two separate numeric columns
    df[["plugin_CI_lower","plugin_CI_upper"]] = pd.DataFrame(
       df["plugin_confidence_interval"]
         .apply(eval)   # turns the string "[-15.3, -14.9]" into a list
         .tolist(),
       index=df.index
    )
    df[["BS_CI_lower","BS_CI_upper"]] = pd.DataFrame(
       df["BS_confidence_interval"]
         .apply(eval)
         .tolist(),
       index=df.index
    )

    # preserve ground‐truth and file id
    df["value_GT"] = df["value_GT"].iloc[0]
    df["batch"]    = fn  # if you want to keep track of which file it came from

    all_loops.append(df[[
      "batch","value_GT",
      "v_pi_fqe","plugin_sigma_squared_avg",
      "plugin_interval_width","plugin_CI_lower","plugin_CI_upper",
      "BS_variance","BS_interval_width","BS_CI_lower","BS_CI_upper"
    ]])

# now stack them all
big_df = pd.concat(all_loops, ignore_index=True)

# for plugin
plugin_cover_flag = (
    (big_df["plugin_CI_lower"] <= big_df["value_GT"]) &
    (big_df["plugin_CI_upper"] >= big_df["value_GT"])
)
total_plugin_covered = plugin_cover_flag.sum()
total_loops            = len(big_df)
overall_plugin_cov     = total_plugin_covered / total_loops

# for bootstrap
BS_cover_flag = (
    (big_df["BS_CI_lower"] <= big_df["value_GT"]) &
    (big_df["BS_CI_upper"] >= big_df["value_GT"])
)
total_BS_covered = BS_cover_flag.sum()
overall_BS_cov   = total_BS_covered / total_loops

print(f"Plugin:  covered {total_plugin_covered}/{total_loops} = {overall_plugin_cov:.3f}")
print(f"Bootstrap: covered {total_BS_covered}/{total_loops} = {overall_BS_cov:.3f}")

total_loops          = len(big_df)
total_plugin_covered = plugin_cover_flag.sum()
overall_plugin_cov   = total_plugin_covered / total_loops
total_BS_covered     = BS_cover_flag.sum()
overall_BS_cov       = total_BS_covered / total_loops
N = len(big_df)

gt = big_df["value_GT"].iloc[0]
big_df["value_GT"] = [gt] + [None] * (N - 1)

big_df["total_loops"]            = [ total_loops            ] + [None] * (N - 1)
big_df["total_plugin_covered"]   = [ total_plugin_covered   ] + [None] * (N - 1)
big_df["overall_plugin_cov"]     = [ overall_plugin_cov     ] + [None] * (N - 1)
big_df["total_BS_covered"]       = [ total_BS_covered       ] + [None] * (N - 1)
big_df["overall_BS_cov"]         = [ overall_BS_cov         ] + [None] * (N - 1)

big_df.to_csv("plugin_BS_CIlevel_0.90_K_900_Hbar_18.0_H_15_transition_0.010_target_0.000_behavior_0.300_B_100_subsetsize_210_all.csv", index=False)

```
